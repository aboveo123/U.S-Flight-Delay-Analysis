```python
import matplotlib.pyplot as plt
import pandas as pd

# Thiết lập dữ liệu từ các bảng thiết kế
data_a = {
    'Timestamp': pd.to_datetime(['2024-01-01', '2024-03-15', '2024-06-01', '2024-08-20', '2024-11-05', '2025-01-10', '2025-04-01', '2025-06-15']),
    'Price': [10.00, 12.00, 12.00, 15.00, 15.00, 18.00, 18.00, 22.00]
}
df_a = pd.DataFrame(data_a)

data_b = {
    'Timestamp': pd.to_datetime(['2024-01-15', '2024-03-01', '2024-05-15', '2024-07-01', '2024-09-15', '2024-11-01', '2024-12-15', '2025-02-01']),
    'Budget': [0.00, 5000.00, 12000.00, 8000.00, 15000.00, 20000.00, 10000.00, 25000.00]
}
df_b = pd.DataFrame(data_b)

# Khởi tạo khung vẽ chứa 4 biểu đồ độc lập
fig, axs = plt.subplots(2, 2, figsize=(14, 10))

# Trực quan hóa Tập dữ liệu A
axs.step(df_a, df_a['Price'], where='post', color='green', marker='o', label="Cấu hình ĐÚNG: step='post'")
axs.set_title("Tập dữ liệu A (SaaS Price) - Hiển thị Chính xác")
axs.set_ylabel("Đơn giá (USD)")
axs.grid(True, linestyle=':')
axs.legend()

axs.step(df_a, df_a['Price'], where='pre', color='red', marker='o', label="Cấu hình SAI: step='pre'")
axs.set_title("Tập dữ liệu A (SaaS Price) - Hiển thị Sai lệch")
axs.set_ylabel("Đơn giá (USD)")
axs.grid(True, linestyle=':')
axs.legend()

# Trực quan hóa Tập dữ liệu B
axs.step(df_b, df_b, where='pre', color='green', marker='o', label="Cấu hình ĐÚNG: step='pre'")
axs.set_title("Tập dữ liệu B (Project Budget) - Hiển thị Chính xác")
axs.set_ylabel("Ngân sách (USD)")
axs.grid(True, linestyle=':')
axs.legend()

axs.step(df_b, df_b, where='post', color='red', marker='o', label="Cấu hình SAI: step='post'")
axs.set_title("Tập dữ liệu B (Project Budget) - Hiển thị Sai lệch")
axs.set_ylabel("Ngân sách (USD)")
axs.grid(True, linestyle=':')
axs.legend()

plt.tight_layout()
plt.savefig('step_charts_output.png', dpi=300)
plt.close()
```
